# normalize the data

````python
X_train=X_train/255
X_test=X_test/255
````

# build the model

````python
from tensorflow.keras import datasets,layers,models
model=models.Sequential()

model.add(layers.Conv2D(32, #filters
                        (3,3), # filter size / kernel size
                        activation='relu',
                        input_shape=(32,32,3))) #stride/paddding by default


model.add(layers.MaxPool2D((2,2))) # shape of the pool


model.add(layers.Conv2D(64, # filters,
                        (3,3), # filter size/ kernel size
                        activation='relu')) # stride/ padding by default

model.add(layers.MaxPool2D((2,2))) # shape of the pool

model.add(layers.Conv2D(64, # filters
                        (3,3), #filter size / kernel size
                        activation='relu'))

````

````python
model.summary()
````

# add dense layers

````python
model.add(layers.Flatten()) # after convolution ==== fully connect layer
# no need of input shape here
model.add(layers.Dense(64,activation='relu')) # hidden layer 64 neurons, relu activation function
#model.add(layers.dense(10)) # 10 classes so 10 neurins
model.add(layers.Dense(10,activation='softmax')) # softmax activation function
````

````python
model.summary()
````

# compile

````python
from re import VERBOSE
model.compile(optimizer='adam',
              loss = 'sparse_categorical_crossentropy',
              metrics=['accuracy'])
````

# fit the model


````python
from sklearn.model_selection import train_test_split
X_train,X_val,y_train,y_val=train_test_split(X_train,y_train,test_size=0.2)
````

````python
history=model.fit(X_train,
                  y_train,
                  epochs=30,
                  batch_size=64,
                  validation_data=(X_val,y_val))
````

````python
history.history
````

````python
history.history.keys()
````

````python
plt.plot(history.history['accuracy'],label='accuracy')
plt.plot(history.history['val_accuracy'],label='val_accuracy')
````

````python
test_loss,test_acc = model.evaluate(X_test,y_test)
````
